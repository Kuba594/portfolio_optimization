```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis
```

```python

rets = pd.read_csv("data/daily_returns.txt", sep="\t", index_col=0, parse_dates=True)
```

```python
stats = pd.DataFrame({
    "mean": rets.mean(),
    "std": rets.std(),
    "skew": rets.skew(),
    "kurtosis": rets.kurtosis(),
    "min": rets.min(),
    "max": rets.max()
})

print(stats)
```

```python
corr = rets.corr()

avg_corr = (corr.values[np.triu_indices_from(corr.values, k=1)]).mean()
min_corr = (corr.values[np.triu_indices_from(corr.values, k=1)]).min()
max_corr = (corr.values[np.triu_indices_from(corr.values, k=1)]).max()

avg_corr, min_corr, max_corr
```
